```python
from sklearn.preprocessing import OneHotEncoder
import numpy
```

```python
documents=["people watch movie",
           "people watch cricket",
           "people like movie",
           "people like cricket",
           "people watch movie"]
```

```python
tokens=[sentence.lower().split() for sentence in documents]
print(tokens)
```

```python
all_words= [[word] for sentence in tokens for word in sentence]
print(all_words)
```

```python
oneHot_encoder=OneHotEncoder(sparse_output=False)
oneHot_encoded=oneHot_encoder.fit(all_words)
print(oneHot_encoded.categories_)
```

```python
print(oneHot_encoded.categories_[0])
```

```python
for sentence in tokens:
    oneHot_encoded_sentence=oneHot_encoder.transform([[word] for word in sentence])
    print(sentence)
    print(oneHot_encoded_sentence)
```
