OneHotEncoder

```python
from sklearn.preprocessing import OneHotEncoder
import numpy
```

```python
documents=["people watch movie",
           "people watch cricket",
           "people like movie",
           "people like cricket",
           "people watch movie"]
```

```python
tokens=[sentence.lower().split() for sentence in documents]
print(tokens)
```

```python
all_words= [[word] for sentence in tokens for word in sentence]
print(all_words)
```

```python
oneHot_encoder=OneHotEncoder(sparse_output=False)
oneHot_encoded=oneHot_encoder.fit(all_words)
print(oneHot_encoded.categories_)
```

```python
print(oneHot_encoded.categories_[0])
```

```python
for sentence in tokens:
    oneHot_encoded_sentence=oneHot_encoder.transform([[word] for word in sentence])
    print(sentence)
    print(oneHot_encoded_sentence)
```

TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents=["people watch movie and people watch cricket",
           "people watch cricket and people like movie",
              "people like movie and people like cricket",
              "people like cricket and people watch movie",
                "people watch movie"]
```

```python
tf_idf=TfidfVectorizer()
```

```python
tf_idf_vector=tf_idf.fit_transform(documents)
```

```python
tf_idf_vector
```

```python
tf_idf_vector.toarray()
```

```python
for vector in tf_idf_vector.toarray():
    print(vector)
```

```python
tf_idf.get_feature_names_out()
```

```python
print("vocabulary:", tf_idf.get_feature_names_out())
```
